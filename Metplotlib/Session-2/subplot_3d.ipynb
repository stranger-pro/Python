```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../Data_set/batter.csv')
df.head()
```

```python
# for four use 0,0 1,0 0,1 1,1 with axis

fig , axs = plt.subplots(ncols=2,nrows=1,figsize=(15,10))
axs[0].scatter(df['avg'],df['runs'])
axs[1].hist(df['runs'])
fig.show()
```

```python
fig = plt.figure()

axs = fig.add_subplot(111,projection='3d')

x=[1,2,3,4]
y = [2,3,4,5]
z = [3,4,5,6]

axs.scatter(x,y,z,marker='+',c='red')
axs.plot3D(x,y,z,marker='+')
```

```python
fig = plt.figure(figsize=(15,9))
ax = fig.add_subplot(projection='3d')
x = np.linspace(30,-30,100)
y = np.linspace(30,-30,100)
x,y = np.meshgrid(x,y)
z=x**2 + y**2
p = ax.plot_surface(x,y,z,cmap='winter')
fig.colorbar(p)
```

```python
fig = plt.figure()
ax = fig.add_subplot()

p = ax.contour(x,y,z)
fig.colorbar(p)
```

```python
fig = plt.figure()
ax = fig.add_subplot()

p = ax.contourf(x,y,z)
fig.colorbar(p)
```
