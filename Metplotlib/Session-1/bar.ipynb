```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as s
```

```python
data = pd.read_csv('../Data_set/batsman_season_record.csv')
data
```

```python
plt.bar(data['batsman'],data['2015'],width=0.4,color='orange')
plt.xticks(rotation='vertical')
plt.show()
```

```python
plt.barh(data['batsman'],data['2015'])
plt.show()
```

```python
plt.bar(data['batsman'],data['2015'],label='2015')
plt.bar(data['batsman'],data['2016'],bottom=data['2015'],label='2016')
plt.bar(data['batsman'],data['2017'],bottom=(data['2015']+data['2016']),label='2017')
plt.legend()
plt.show()
```
