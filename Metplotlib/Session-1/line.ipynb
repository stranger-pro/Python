```python
data = pd.read_csv('../Data_set/sharma-kohli.csv')

Rohit = data['RG Sharma']
Kohli = data['V Kohli']
```

```python
index = data['index']

plt.plot(index,Kohli,color='green')
plt.plot(index,Rohit,color='black',linestyle='dashdot')
plt.title('Rohit VS Kohli')
plt.xlabel('Year',labelpad=15)
plt.ylabel('Run',labelpad=15)
```

```python
index = data['index']

plt.plot(index,Kohli,color='green',linestyle='dashed',linewidth=3,marker='D',label='Kohli')
plt.plot(index,Rohit,color='black',linewidth=2,marker='o',label='Rohit')
plt.title('Rohit VS Kohli')
plt.xlabel('Year',labelpad=15)
plt.ylabel('Run',labelpad=15)
plt.legend()
plt.show()
```
