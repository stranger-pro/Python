```python
df1 = pd.read_csv('../Data_set/iris.csv')
df = df1.sample(40)

df['Species'] =  df['Species'].replace({'Iris-setosa':0,
                       'Iris-versicolor':1,
                       'Iris-virginica':2})
```

```python
plt.figure(figsize=(10,8))
plt.scatter(df['SepalLengthCm'],df['SepalWidthCm'],c=df['Species'],cmap='winter')
plt.xlabel('SepalLengthCm',labelpad=15)
plt.ylabel('SepalWidthCm',labelpad=15)
plt.colorbar()
plt.show()
```

```python
plt.figure(figsize=(10,8))
plt.scatter(df['SepalLengthCm'],df['SepalWidthCm'],c=df['Species'],cmap='winter')
plt.xlabel('SepalLengthCm',labelpad=15)
plt.ylabel('SepalWidthCm',labelpad=15)
for i in range(df.shape[0]):
    plt.text(df['SepalLengthCm'].values[i],df['SepalWidthCm'].values[i],f'point_{i}')
plt.colorbar()
plt.show()
```
