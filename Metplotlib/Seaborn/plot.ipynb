```python
import matplotlib as plt
import seaborn as sns
import plotly.express as px
```

```python
df = sns.load_dataset('tips')
```

```python
df.head(3)
```

```python
sns.scatterplot(df,x='tip',y='total_bill',hue='sex',style='time',size='size')
```

```python
sns.relplot(df,x='total_bill',y='tip',kind='scatter',hue='sex',style='smoker',size='size')
```

```python
# this work only figure level function only
sns.relplot(df,x='total_bill',y='tip',kind='scatter',row='sex',col='smoker',hue='time')
```

```python
sns.lineplot(df,x='total_bill',y='size',hue='sex',style='smoker',size='size')
```

```python
sns.relplot(df,x='total_bill',y='size',kind='line',col='size',col_wrap=3)
```
