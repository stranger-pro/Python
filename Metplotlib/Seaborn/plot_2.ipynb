```python
import matplotlib as plt
import seaborn as sns
import plotly.express as px
```

```python
# facet is available for all figure level
```

```python
df = sns.load_dataset('tips') 
```

```python
df.head()
```

```python
sns.stripplot(df,x='day',y='tip',hue='day')
```

```python
sns.catplot(df,x='day',y='total_bill',kind='strip',hue='day')
```

```python
sns.catplot(df,x='day',y='total_bill',kind='swarm',hue='day')
```

```python
sns.boxplot(df,x='day',y='tip',hue='sex')
```

```python
sns.catplot(df,x='day',y='tip',hue='sex',kind='box')
```

```python
sns.boxplot(df,y='tip',hue='sex')
```

```python
sns.violinplot(df,x='day',y='tip',hue='sex')
```

```python
sns.violinplot(df,x='day',y='tip',hue='sex',split=True)
```

```python
sns.barplot(df,x='sex',y='tip',hue='day')
```

```python
sns.barplot(df,x='sex',y='tip',hue='day',estimator='min',errorbar=None)
```

```python
sns.countplot(df,x='sex',hue='day')
```

```python
sns.pointplot(df,x='sex',y='tip',hue='day')
```

```python
sns.lmplot(df,x='tip',y='total_bill',hue='sex')
```

```python
sns.residplot(df,x='tip',y='total_bill')
```
