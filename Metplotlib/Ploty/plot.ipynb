```python
import numpy as np
import pandas as pd
import plotly.express as px
```

```python
data = px.data.gapminder()
data.head()
```

```python
temp = data[data['year']==2007].sample(30)
```

```python
px.scatter(data,x='lifeExp',y='pop',color='country',size='iso_num')
```

```python
px.scatter(temp,x='lifeExp',y='pop',color='country',size='iso_num')
```

```python
px.scatter(data,x='lifeExp',y='gdpPercap',color='continent',size='pop',
           size_max=50,range_x=[0,85],
           animation_frame='year',animation_group='country')
```

```python
df = data[data['country'] =='India']
```

```python
px.line(df,x='year',y='pop',markers='o')
```

```python
temp_df = data[data['country'].isin(['India','United States','China'])].pivot(index='year',columns='country',values='lifeExp')
temp_df
```

```python
px.line(temp_df,x=temp_df.index,y=temp_df.columns)
```

```python
temp_df = data[data['country'].isin(['India','United States','China'])].pivot(index='year',columns='country',values='gdpPercap')
temp_df
```

```python
px.bar(temp_df,x=temp_df.index,y=temp_df.columns,barmode='group',text_auto=True,log_y=True)
```
