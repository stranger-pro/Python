```python
import numpy as np

arr = np.arange(16).reshape(2,2,2,2)

print(arr)
```

```python
import numpy as np

arr = np.arange(16).reshape(2,2,2,2)

print(arr.size)
```

```python
import numpy as np

arr = np.arange(16).reshape(2,2,2,2)

arr1 = np.resize(arr, (4,4))

print(arr.shape)
print(arr1.shape)
```

```python
import numpy as np

arr = np.arange(16,dtype=np.int64).reshape(2,2,2,2)

arr1 = np.resize(arr, (4,4))

print(arr.itemsize)

print(arr1.itemsize)

print(arr.dtype)

print(arr1.dtype)
```
