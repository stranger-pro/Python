```python
# Sigmoid function

import numpy as np
def sigmoid(a):
    return 1/(1 + np.exp(-(a)))

aar = np.arange(10)
sigmoid(aar)
```

```python
# Mean Square Error

import numpy as n
import random

def mse(actual , pridiction):
    return n.mean(((actual - pridiction)**2))

actual = n.random.randint(1,50,20)
pridiction = n.random.randint(1,50,20)

print(mse(actual,pridiction))
```
