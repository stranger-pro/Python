```python
import numpy as np 

arr = np.array([1,2,3,22,4,3,6,1,7,9,8,3,19,5,6])

np.unique(arr)
```

```python
# Where is the max value

print(np.argmax(arr))

np.argmin(arr)
```

```python
# Where provide you index of elelments

np.where(arr>4)
```

```python
np.histogram(arr,bins=[0,10,20,30])
```

```python
# Replace values with 0
arr = np.where(arr>4,0,arr)
```
