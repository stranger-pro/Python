```python
import numpy as np
import random

arr = np.random.randint(1,100,20)

# Sort

print(np.sort(arr))

#reverse

print(np.sort(arr)[::-1])
```

```python
import numpy as np
import random

arr = np.random.randint(1,100,20).reshape(4,5)

print(np.sort(arr,axis=0))  # for column axis=0 and for row axis=1

print(arr.shape)
```

```python
import numpy as np

import random

arr = np.random.randint(1,100,20).reshape(4,5)

arr1 = np.append(arr,np.ones((1,arr.shape[1])),axis=0)

print(arr1)
```

```python
import numpy as np

import random

arr = np.random.randint(1,100,20).reshape(4,5)
arr1 = np.random.randint(100,200,20).reshape(4,5)

print(np.concatenate((arr,arr1),axis=1))
```
