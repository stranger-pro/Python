```python
import numpy as n
import matplotlib.pyplot as m
import random as random 

y = n.random.randint(1,100,size=20)
x = n.linspace(0,100,20)
m.title('Satta Bazzer',pad=20)
m.xlabel('Investment',labelpad=20)
m.ylabel('Return',labelpad=20)
m.bar(x,y)

m.show()
```

```python
# Sigmoid function Graph

import numpy as np
import matplotlib.pyplot as m

def sigmoid(a):
    return 1/(1 + np.exp(-(a)))

aar = np.linspace(-10,10,100)
print(m.plot(sigmoid(aar)))
```
