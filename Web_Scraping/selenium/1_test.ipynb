```python
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import numpy as np
from bs4 import BeautifulSoup
import pandas as pd
```

```python
driver = webdriver.Edge()
driver.get("https://www.smartprix.com/laptops")
time.sleep(3)
link = driver.find_element(By.XPATH,'//*[@id="app"]/main/aside/div/div[3]/div[3]/label[1]/input')
link.click()

time.sleep(3)

i=0
old_size = driver.execute_script('return document.body.scrollHeight')
while True:
    load = driver.find_element(By.XPATH,'//*[@id="app"]/main/div[1]/div[2]/div[3]')

    try:
        load.click()

        time.sleep(5)
        new_size = driver.execute_script('return document.body.scrollHeight')

        if((new_size == old_size)&(i>100)):
            break

        old_size=new_size
        i=i+1
        print(new_size , "  " , i)

    except:
        print("Not Click")

    

html_text = driver.page_source

with open('data.html','w',encoding='utf-8') as f:
    f.write(html_text)
```

```python
with open('data.html','r',encoding='utf*') as f:
    r = f.read()
```

```python
soup = BeautifulSoup(r,'html')
```

```python
text = soup.find_all('div',{'class':"sm-product"})

```

```python
name=[]
price = []
rating=[]
proccesor = []
core = []
thread = []
ram =[]
storage=[]
Graphic_card=[]
display=[]
pixel=[]
os=[]
for i in text:
    # name

    try:
        name.append(((str(i.find('h2'))).split('>')[1]).split('(')[0].strip())
    except:
        name.append(np.nan)

    # price

    try:
        price.append(((str(text[0].find('span').text)).replace('₹','')).replace(',',''))
    except:
        price.append(np.nan)

    # rating

    try:
        rating.append(int(i.find('div',{"class":'score rank-2-bg'}).text[:2]))
    except:
        rating.append(np.nan)

    # proccesor

    try:
        proccesor.append(((str(i.find('ul',{"class":'sm-feat specs'}))).split('>')[2]).split('<')[0])
    except:
        proccesor.append(np.nan)
    

    # core

    try:
        core.append(((str(i.find('ul',{"class":'sm-feat specs'}))).split('>')[4]).split(',')[0])
    except:
        core.append(np.nan)

    # thread
    
    try:
        thread.append((((str(i.find('ul',{"class":'sm-feat specs'}))).split('>')[4]).split(',')[1]).split(' ')[1])
    except:
        thread.append(np.nan)

    #ram
    
    try:
        ram.append(((str(i.find('ul',{"class":'sm-feat specs'}))).split('>')[6]).split(' ')[0][:2])
    except:
        ram.append(np.nan)

    # storage

    try:
       storage.append((((str(i.find('ul',{"class":'sm-feat specs'}))).split('>')[8]).split(" ")[0]).replace('\u2009',''))
    except:
        storage.append(np.nan)

    # Graphic_card

    try:
        Graphic_card.append(((str(i.find('ul',{"class":'sm-feat specs'}))).split('>')[10]).split('<')[0].replace('\u2009',''))
    except:
        Graphic_card.append(np.nan)

    # display

    try:
        display.append(i.find('ul',{"class":'sm-feat specs'}).find_all('li')[5].text.split(',')[0].replace('\u2009',' '))
    except:
        display.append(np.nan)

    # pixel

    try:
        pixel.append(i.find('ul',{"class":'sm-feat specs'}).find_all('li')[5].text.split(',')[1].replace('\u2009',' '))
    except:
        pixel.append(np.nan)

    # os

    try:
        t = i.find('ul',{"class":'sm-feat specs'}).find_all('li')[-2].text
        if(t[-8::1]=='Warranty'):
            os.append(np.nan)
        else:
            os.append(t)
            
    except:
        os.append(np.nan)
```

```python
HP = pd.DataFrame({'name':name,
'price' : price,
'rating':rating,
'proccesor' : proccesor,
'core':core,
'thread' :thread,
'ram':ram,
'storage':storage,
'Graphic_card':Graphic_card,
'display':display,
'pixel':pixel,
'os':os})
```

```python
t = HP[HP['name'].isnull()]

HP.drop(index=t.index,inplace=True)
```
