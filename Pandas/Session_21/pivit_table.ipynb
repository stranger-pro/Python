```python
data = pd.read_csv('../Data_set/expense_data.csv')
data
```

```python
data['Date'] = pd.to_datetime(data['Date'])
```

```python
data['Month']=data['Date'].dt.month_name()
data.head()
```

```python
data.pivot_table(index=['Month'],columns=['Category'],values=['Amount'])
```

```python
data.pivot_table(index=['Month'],columns=['Category'],values=['Amount']).plot()
```

```python
data.pivot_table(index=['Month'],columns=['Income/Expense'],values=['Amount'],aggfunc='sum').plot()
```
