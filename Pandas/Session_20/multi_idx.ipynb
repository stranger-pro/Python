```python
data = pd.DataFrame(
    [[1,2,3,4,5,6],
     [11,12,13,14,15,16],
     [21,22,23,24,25,26],
     [32,32,33,34,35,36]
     ],

     index=pd.MultiIndex.from_product([['ITM','BUDDHA'],['SGPA','CGPA']]),
     columns=pd.MultiIndex.from_tuples(multi_idx)
)
```

```python
data 
```

```python
data.unstack() #use unstack function to change row into column
```

```python
data.unstack().stack() #use stack function to change column into row
```

```python
data.sort_index(ascending=False) #first level
```

```python
data.sort_index(level=1,ascending=False) #level second
```

```python
data.sort_index(ascending=[False,True]) # both level
```
